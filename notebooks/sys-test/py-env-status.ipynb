## python venv upgrade

### venv direct update

```bash

# ------------------------------------------------------

(type nvidia-smi && nvidia-smi -L) &> /dev/null \
    && X_HAS_GPU=1 || X_HAS_GPU=0;

case "$X_HAS_GPU" in
     1) export X_UV_EXTRA='gpu' ;;
     *) export X_UV_EXTRA='cpu' ;;
esac
export X_HAS_GPU

# ------------------------------------------------------



uv lock
uv sync --extra=$X_UV_EXTRA --all-groups

```
