```python
from sympy import init_printing, symbols, Function, Derivative, Integral, Eq, integrate, diff, dsolve, checkodesol, oo, exp, sin
from sympy.abc import x
from spb import plot

init_printing()

```

## Integral

```python
x, y, z, t = symbols("x y z t")
```

```python
f = exp(-(x**2) - y**2)
F = f , (x, -oo, oo), (y, -oo, oo)
```

```python
Integral(*F)
```

```python
integrate(*F)
```

## Derivative

```python
g = sin(x) / x
```

```python
dg = diff(g)
```

```python
g
```

```python
dg
```

```python
plot(g, dg)
```

## Differential Equation (ODE)

```python
y = Function("y")
eq = Derivative(y(x), x, x) + 9*y(x)
Eq(eq,0)
```

```python
sol = dsolve(eq, y(x))
sol
```

```python
checkodesol(eq, sol)
```
